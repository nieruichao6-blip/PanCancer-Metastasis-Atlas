```python
import os
import gc
import certifi
import pandas as pd
import numpy as np
import scanpy as sc
import diopy
import bbknn
import scvi
import scrublet as scr
import cellrank as cr
import matplotlib.pyplot as plt
import scanpy.external as sce
import scvelo as scv
import scib
from cellrank.kernels import PseudotimeKernel
from cellrank.kernels import CytoTRACEKernel
from scib_metrics.benchmark import Benchmarker


sc.settings.set_figure_params(dpi=3000,figsize=(5, 5))
sc.logging.print_header()
```

```python
# data dir
input_dir = " "
output_dir = " "

# cell color
cell_type_color = {"Epithelia":"#A6CEE3","Fibroblast":"#FDBF6F","Endothelia":"#B2DF8A","Acinar_cell":"#984EA3","SMC&Pericyte":"#33A02C","T_cell":"#1F78B4","NK_cell":"#64abc0","B_cell":"#FF7F00",
                   "Plasma_cell":"#FB9A99","Myeloid_cell":"#CAB2D6","Mast_cell":"#6A3D9A","Neutrophils":"#FFFF99","Hepatocyte":"#9ec9e1","Melanoma_cell":"#E66F00","Neuron":"#1d92c0",
                   "Glial_cell":"#B15928","Osteoblastic_cell":"#fcc5c1","Alpha_cell":"#42aa5e","Bela_cell":"#c22b86","undefine":"#A6CEE3"}
tumor_status_color = {"Tumor":"#5BC0EB","Metastasis":"#8DD3C7","NAT":"#9BC53D","Healthy":"#C3423F"}

os.listdir(input_dir)
```

```python
# plot run
for file in os.listdir(input_dir):
    print(f"############################# plot {file} process ##########################################")

    ## output dir
    output_file = f"{output_dir}/{file}"

    ## data read
    scRNA_current = diopy.input.read_h5(file = f"{input_dir}/{file}/scRNA_annotation.h5")

    ## data plot
    sc.pl.umap(scRNA_current, color='cell_type',show = False,legend_loc=None,palette = cell_type_color)
    plt.savefig(f'{output_file}/UMAP_annotation_nolabel_scanpy.png', dpi=3000)
    plt.close()
    sc.pl.umap(scRNA_current, color='cell_type',show = False,legend_loc='on data',palette = cell_type_color)
    plt.savefig(f'{output_file}/UMAP_annotation_scanpy.png', dpi=3000)
    plt.close()
    sc.pl.umap(scRNA_current, color='tumor_status',show = False,legend_loc=None,palette = tumor_status_color)
    plt.savefig(f'{output_file}/UMAP_tumor_status_scanpy.png', dpi=3000)
    plt.close()
    sc.pl.umap(scRNA_current, color='sample_ID',show = False,legend_loc=None)
    plt.savefig(f'{output_file}/UMAP_sample_distribution_scanpy.png', dpi=3000)
    plt.close()

    ## space free
    del scRNA_current
    gc.collect()
    
```
