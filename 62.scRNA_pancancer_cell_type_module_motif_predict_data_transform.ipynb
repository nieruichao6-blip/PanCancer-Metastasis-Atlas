```python
import os
import gc
import pandas as pd
import numpy as np
import scanpy as sc
import diopy
import bbknn
import scrublet as scr
import matplotlib.pyplot as plt
import scanpy.external as sce
import hotspot
import loompy as lp

sc.settings.set_figure_params(dpi=1000,figsize=(5, 5))
sc.logging.print_header()
```

```python
## data dir
input_dir = " "
output_dir = " "

# file_list = ["scRNA_CM1_count","scRNA_CM2_count","scRNA_CM3_count","scRNA_CM4_count","scRNA_CM5_count","scRNA_CM6_count","scRNA_CM7_count","scRNA_CM8_count","scRNA_CM9_count"]
file_list = ["scRNA_CM5_count","scRNA_CM6_count","scRNA_CM7_count","scRNA_CM8_count","scRNA_CM9_count"]
```

```python
## data process
for file in file_list:
    print(f"---------------  {file}")

    ### data read
    counts = pd.read_csv(f"{input_dir}/{file}.csv",index_col = 0, header = 0)

    ### data code
    barcodes = counts.index.tolist()
    features = counts.columns.tolist()

    ### loom data save
    row_attrs = {'Gene':np.array(features)}
    col_attrs = {'CellID':np.array(barcodes)}
    lp.create(f"{output_dir}/{file}.loom", counts.values.transpose(), row_attrs, col_attrs)
```
