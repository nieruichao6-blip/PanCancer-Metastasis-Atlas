```python
import os
import pandas as pd
import numpy as np
import scanpy as sc
import diopy
import infercnvpy as cnv
import matplotlib.pyplot as plt
import gc


sc.settings.set_figure_params(dpi=300,figsize=(5, 5))
sc.logging.print_header()
```

```python
# data dir
input_dir = " "
# input_dir = " "
output_dir = " "
gencode_ref_dir = " "

# data read
# scRNA_current = sc.read_h5ad(f"{input_dir}/scRNA_annotation_tumor_transform.h5ad")
scRNA_current = sc.read_h5ad(f"{input_dir}/scRNA_gene_filter.h5ad")
```

```python
# infercnv analysis
## chr ano
cnv.io.genomic_position_from_gtf(adata=scRNA_current, gtf_file=gencode_ref_dir,gtf_gene_id='gene_name', inplace=True)

## infercnv analysis
cnv.tl.infercnv(
    scRNA_current,
    # reference_key="cell_type",
    # reference_cat=reference_cell_type,
    window_size=100,
    n_jobs=24
    )

## cnv plot
os.chdir(output_dir)
cnv.pl.chromosome_heatmap(scRNA_current, groupby="new_cell_type",save="scRNA_chromosome_cell_type.pdf")
cnv.pl.chromosome_heatmap(scRNA_current, groupby="tumor_status",save="scRNA_chromosome_tumor_status.pdf")
cnv.pl.chromosome_heatmap(scRNA_current, groupby="tumor_code",save="scRNA_chromosome_tumor_code.pdf")

## data save
diopy.output.write_h5(scRNA_current, file = f"{output_dir}/scRNA_tumor_cell_infercnv.h5",save_X=False)
scRNA_current.write_h5ad(f"{output_dir}/scRNA_tumor_cell_infercnv.h5ad", compression="gzip")
```
