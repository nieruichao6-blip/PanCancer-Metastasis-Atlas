```python
## data dir
input_dir = " "
output_dir = " "
```

```python
## data read
# counts = pd.read_csv(f"{input_dir}/scRNA_tumor_cell_counts_part3.csv",index_col = 0, header = 0)
counts = pd.read_csv(f"{input_dir}/scRNA_tumor_cell_counts_test.csv",index_col = 0, header = 0)
```

```python
barcodes = counts.index.tolist()
features = counts.columns.tolist()
```

```python
## loom data save
row_attrs = {'Gene':np.array(features)}
col_attrs = {'CellID':np.array(barcodes)}
lp.create(f"{output_dir}/counts_test.loom", counts.values.transpose(), row_attrs, col_attrs)
# lp.create(f"{output_dir}/counts_part3.loom", counts.values.transpose(), row_attrs, col_attrs)
```
